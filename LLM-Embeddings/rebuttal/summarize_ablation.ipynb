```python
import json
import numpy as np
```

```python
fname = '../results/google-bert_bert-base-uncased_fmsr_ft_bs32_poolmean_inbatchnegs_pdesc0.4_positiveratio_0.5seed_<seed>.json'
```

```python
fname_seed = fname.replace('<seed>', '42')
with open(fname_seed, 'r') as f:
    txt = f.read()
    results = json.loads(txt.split('\n')[0])
all_tasks = set()
for key in results:
    task = key.split('_test_')[0]
    all_tasks.add(task)
all_tasks.remove('sequential_score_before')
all_tasks.remove('sequential_score_after')
suffixes = ['test_cosine_accuracy@1', 'test_cosine_map@100', 'test_cosine_ndcg@10']
full_names = []
for task in all_tasks:
    for suffix in suffixes:
        full_names.append(task + '_' + suffix + '_after')
```

```python
stats = {name: [] for name in full_names}
for seed in range(42, 47):
    fname_seed = fname.replace('<seed>', str(seed))
    with open(fname_seed, 'r') as f:
        txt = f.read()
        results = json.loads(txt.split('\n')[0])
        for key in full_names:
            stats[key].append(results[key])
```

```python
mapping = {
    'eq_to_category': 'E2CAT',
    'asset_fault_to_sensor': 'FM2S',
    'failure_desc_to_class': 'FM2CLS',
    'eq_to_class_type': 'E2CLT',
    'asset_sensor_to_fault': 'S2FM',
    'asset_fm_to_components': 'FM2C',
    'eq_subunit_to_unit': 'EU2SU',
    'component_to_failure_mode': 'C2FM',
    'asset_to_related_sensors': 'A2S'
}
metric_mapping = {
    'accuracy@1': 'HIT@1',
    'map@100': 'MAP@100',
    'ndcg@10': 'NDCG@10'
}
seen = set()
for key in stats:
    for task in mapping:
        if task in key and not mapping[task] in seen:
            print('\n'+ mapping[task])
            seen.add(mapping[task])
    for metric in metric_mapping:
        if metric in key:
            mean = round(np.mean(stats[key]), 2)
            std = round(np.std(stats[key]), 2)
            print(metric_mapping[metric], mean, std)
```
